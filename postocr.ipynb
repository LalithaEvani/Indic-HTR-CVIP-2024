```python
import lmdb
from tqdm import tqdm
```

```python
def create_lexicon(folder_path):
    env_data = lmdb.open(folder_path, readonly=True)
    txn_data = env_data.begin()
    cursor_data = txt_data.cursor()

    lexicon_list = []
    for key, value in tqdm(cursor_data, desc='saving images'):
        if key.startswith(b'label-'):
            label_value = value.decode().strip()
            
            if label_value not in lexicon_list:
                lexicon_list.append(label_value)
    return lexicon_list
```

```python
lexicon_list = []
train_folder_path = 'data/train/real/hindi'
val_folder_path = 'data/val/hindi'
test_folder_path = 'data/test/hindi'

lexicon_list.extend(create_lexicon(train_folder_path))
lexicon_list.extend(create_lexicon(val_folder_path))
lexicon_list.extend(create_lexicon(test_folder_path))

lexicon_list.shape

```
